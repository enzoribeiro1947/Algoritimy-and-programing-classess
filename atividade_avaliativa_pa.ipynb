```python
#3

num = int(input("digite um número inteiro de 1 a 10 e veja a sua tabuada a seguir:"))

while num > 10 or num < 10:
  num = int(input("digite um número inteiro de 1 a 10 e veja a sua tabuada a seguir:"))

for n in range(1, 11):
  print(f"{n} X {num} = {n * num}")
```

```python
#4

num = int(input("digite um número: "))

factorial = 1
for n in range(1, num + 1):
  factorial *= n

print(f"O fatorial de {num} é: {factorial}")
```

```python
#5

valor = int(input("qual o valor de pagamento? "))

count100 = 0
count50 = 0
count20 = 0
count10 = 0
count5 = 0
count2 = 0
count1 = 0

while valor >= 100:
    count100 += 1
    valor -= 100

while valor >= 50:
    count50 += 1
    valor -= 50

while valor >= 20:
    count20 += 1
    valor -= 20

while valor >= 10:
    count10 += 1
    valor -= 10

while valor >= 5:
    count5 += 1
    valor -= 5

while valor >= 2:
    count2 += 1
    valor -= 2

while valor >= 1:
    count1 += 1
    valor -= 1

if count100 > 0:
    print(f"{count100} nota(s) de R$100")
if count50 > 0:
    print(f"{count50} nota(s) de R$50")
if count20 > 0:
    print(f"{count20} nota(s) de R$20")
if count10 > 0:
    print(f"{count10} nota(s) de R$10")
if count5 > 0:
    print(f"{count5} nota(s) de R$5")
if count2 > 0:
    print(f"{count2} nota(s) de R$2")
if count1 > 0:
    print(f"{count1} nota(s) de R$1")
```
