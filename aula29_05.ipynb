```python
import random

cont = 0
num = random.randint(1, 10)

while cont < 3:
    chute = int(input("Digite um número de 1 a 10: "))
    cont += 1

    if chute == num:
        print("Parabéns, você acertou o número!")
        break
    else:
        print("Você errou.")

        if cont < 3:
            continuar = input("Deseja continuar? (y/n): ").lower()

            while continuar not in ["y", "n"]:
                continuar = input("Digite apenas y ou n: ").lower()

            if continuar == "n":
                print("Programa fechado.")
                break

else:
    print(f"Suas tentativas acabaram. O número era {num}.")


```
