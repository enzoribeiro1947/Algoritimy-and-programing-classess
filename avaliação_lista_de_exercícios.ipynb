```python
#1

num1 = float(input("digite um número: "))

if num1 % 2 == 0:
  print(f"{num1} é par")
else:
  print("o seu número é impar")
```

```python
#2

num1 = float(input("digite um número: "))
num2 = float(input("digite outro número: "))

if num1 > num2:
  print(f"{num1} é maior que {num2}")
elif num1 < num2:
  print(f"{num2} é maior que {num1}")
else:
  print(f"{num1} é igual a {num2}")
```

```python
#3

nota = float(input("digite a sua nota: "))

if nota >= 7:
  print("você passou!")
elif nota < 7:
  print("você está de reculperação...")
elif nota < 5:
  print("você está reprovado...")
```

```python
#4

idade = int(input("digite sua idade: "))

if idade <= 12:
  print("você é uma criança!")
elif idade > 12 and idade <= 17:
  print("você é um adolescente!")
elif idade >= 18 and idade < 60:
  print("você é um adulto!")
else:
  print("você é um idoso")
```

```python
#6

num1 = float(input("digite um número: "))

if num1 < 0:
  print(f"{num1} é negativo")
elif num1 == 0:
  print("o seu número é zero")
elif num1 > 0:
  print("o seu número é positivo")
```

```python
#7

num1 = float(input("digite um número: "))
num2 = float(input("digite outro: "))

print(20*"_")
print("escolha qual operação você deseja fazer entre esses números")
print("soma = 1")
print("subtração = 2")
print("multiplicação = 3")
print("divizão = 4")
print(20*"_")

soma = num1 + num2
sub = num1 - num2
mult = num1 * num2
div = num1 / num2


escolha = int(input("?: "))

if escolha == 1:
  print(f"{num1} + {num2} = {soma}")
elif escolha == 2:
  print(f"{num1} - {num2} = {sub}")
elif escolha == 3:
  print(f"{num1} x {num2} = {mult}")
elif escolha == 4:
  print(f"{num1} + {num2} = {div}")
```

```python
#8

num1 = int(input("digite um número de 1 a 7: "))

if num1 == 1:
  print("hoje é domingo")
elif num1 == 2:
  print("hoje é segunda")
elif num1 == 3:
  print("hoje é terça")
elif num1 == 4:
  print("hoje é quarta")
elif num1 == 5:
  print("hoje é quinta")
elif num1 == 6:
  print("hoje é sexta")
elif num1 == 7:
  print("hoje é sabado")
else:
  print("digite um número válido")
```

```python
#10

valor = float(input("Digite o valor da compra: "))

if valor >= 100:
    desconto = valor * 0.10
elif valor >= 50:
    desconto = valor * 0.05
else:
    desconto = 0

total = valor - desconto

print("Desconto:", desconto)
print("Total a pagar:", total)
```
