```python
#loops

for n in range(0, 5):
  print("i don't wanna be alone")
```

```python
#crie um programa que recebe um número e retorna apenas números pares

numero = int(input("digite um número: "))

for n in range(numero):
  if n % 2 == 0:
    print(f"{n} é par")
```

```python
#crie um programa que recebe um número e retorna apenas números impares

numero = int(input("digite um número: "))

for n in range(numero):
  if n % 2 != 0:
    print(f"{n} é impar")
```

```python
#escrever um programa que imprime uma contagem regressiva

num = int(input("digite um número: "))

for n in range(num + 1):
  print(num - n)
```

```python
#inverção de palavras

palavra = str(input("digite uma palavra"))
invercao = ""

for n in (palavra):
  invercao = n + invercao

print(invercao)
```

```python
#crie um código que recebe um numero de 1 a 10 e imprime a tabuada

num = int(input("digite um número inteiro de 1 a 10 e veja a sua tabuada a seguir:"))

while num > 10 or num < 10:
  num = int(input("digite um número inteiro de 1 a 10 e veja a sua tabuada a seguir:"))

for n in range(1, 11):
  print(f"{n} X {num} = {n * num}")
```

```python
#soma os números de 0 a n

num = int(input("digite um número: "))
for n in range(num + 1):
  print(f"{num} + {n} = {num + n}")
```
