```python
#crie um programa que calcula a média de notas 4 bimestres (não permitir notas menores que zero e nem maiores que 10)

#média < 5 = reprovado
#média maior que 5 e menor que 7 = reculperação
#média maior ou igual a 7 = aprovado

b1 = float(input("Digite a nota do primeiro bimestre: "))
if b1 < 0 or b1 > 10:
    print("Resultado inválido")
b2 = float(input("Digite a nota do segundo bimestre: "))
if b2 < 0 or b2 > 10:
    print("Resultado inválido")
b3 = float(input("Digite a nota do terceiro bimestre: "))
if b3 < 0 or b3 > 10:
    print("Resultado inválido")
b4 = float(input("Digite a nota do quarto bimestre: "))
if b4 < 0 or b4 > 10:
    print("Resultado inválido")

soma = b1 + b2 + b3 + b4
media = soma / 4

print("Média:", media)

if media < 5:
    print("Reprovado")
elif media >= 5 and media < 7:
    print("Recuperação")
else:
    print("Aprovado")
```
